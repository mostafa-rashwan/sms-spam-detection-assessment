# EDA - SMS Spam Collection
All logic is imported from `spamdetect`; nothing is duplicated here.

```
from spamdetect.config import load_config
from spamdetect.data import parse, clean
from spamdetect.preprocessing import preprocess
cfg = load_config('../config.yaml')
raw = parse(cfg['paths']['raw_dir'] + '/SMSSpamCollection')
df, stats = clean(raw)
stats
```

```
df['label'].value_counts(normalize=True)
```

```
df['len'] = df['text'].str.len()
df.groupby('label')['len'].describe()
```
